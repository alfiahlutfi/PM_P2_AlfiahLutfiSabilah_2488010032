**C. Langkah Latihan**

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]} # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']] # fitur
y = df['harga']                   # label
```

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), 'Uji:', len(X_test))
```

1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?


Supervised datanya berlabel dan tujuannya untuk prediksi. Sedangkan unsupervised datanya itu tanpa label dan tujuannya untuk menemukan struktur. Sementara reinforcement learning itu ada interaksi dan reward dengan tujuannya untuk memaksimalkan reward.

2. Mengapa pembagian data latih dan uji penting?

Pembagian data latih dan data uji penting untuk mengetahui apakah model bisa bekerja dengan baik pada data yang belum pernah dilihat sebelumnya. Jadi data latih digunakan untuk belajar, sedangkan data uji itu digunakan untuk menguji kemampuan model.

3. Mengapa alur kerja ML disebut iteratif?


Karena prosesnya tidak dilakukan sekali, tapi berulang-ulang mulai dari membuat model, menguji hasilnya, melihat apakah sudah sesuai, lalu memperbaiki modelnya lagi kalau ada kekurangan, dan mengujinya lagi sampai mendapatkan hasil yang sesuai keinginan.
